# Diapositiva 2: Funcion de Perdida (MSE)
Para un modelo lineal, la perdida de error cuadratico medio es:

$$L(\theta) = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$$

El objetivo es encontrar $\theta^*$ que minimice $L(\theta)$.
